Set up a proportion:

$\frac{n1}{d1} = \frac{n2}{d2}$


```python
# Put a zero in for the unknown value
# Setting the unknown value to 0 acts as a placeholder or a "flag." 
# a value of 0 is a simple way to tell the computer: "This is the missing piece I want you to calculate."

n1 = 1
d1 = 2
n2 = 0
d2 = 16

if n1 == 0:
    answer = n2 * d1 / d2
    print("n1 =", answer)

if d1 == 0:
    answer = n1 * d2 / n2
    print("d1 =", answer)

if n2 == 0:
    answer = d2 * n1 / d1
    print("n2 =", answer)

if d2 == 0:
    answer = n2 * d1 / n1
    print("d2 =", answer)
```
